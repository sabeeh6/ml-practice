```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error ,r2_score
```

```python
df=pd.read_csv('/content/normalized_experience_salary.csv')
df
```

```python
df.head()
```

```python
df.shape
```

```python
plt.scatter(df['Experience'],df['Salary'])
plt.xlabel('Experince')
plt.ylabel('Salary')
plt.title('Experience vs Salary')
plt.show()
```

```python
X=df[['Experience']]
Y=df['Salary']
# Split the dataset
X_train ,X_test ,Y_train ,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
# create model
model = LinearRegression()
model
```

```python
model.fit(X_train,Y_train)
```

```python
print("Slope (m):", model.coef_)       # Har 1 saal experience par kitni salary barhti hai
print("Intercept (c):", model.intercept_) # Base salary (0 years experience par)
```

```python
y_pred
y_pred = model.predict(X_test)
comparision = pd.DataFrame({'Actual':Y_test,'Predicted':y_pred})
comparision
```

```python
# calculation of the error
mae =-mean_absolute_error(Y_test,y_pred)
r2 = r2_score(Y_test,y_pred)
print("Mean Absolute Error:",mae)
print("R2 Score:",r2)
```

```python
# plot the regression line
plt.scatter(X_test,Y_test,color='blue')
plt.plot(X_test,y_pred,color='red',linewidth=2)
plt.xlabel('Experience')
plt.ylabel('Salary')
plt.title('Experience vs Salary')
plt.show()
```
