### Handling Null Values
Professionally hum `fillna()` use karte hain missing data ko fill karne ke liye ya `dropna()` delete karne ke liye.

```python
# Method 1: Income ko uske average (mean) se fill karna
# Taake data delete na ho aur analysis kharab na ho
mean_income = df['Income'].mean()
df['Income'] = df['Income'].fillna(mean_income)

print(f"Average Income used for filling: {mean_income}")
print("Null values after filling:", df['Income'].isnull().sum())
```

```python
# Method 2: Agar aap kisi row ko delete karna chahen jisme koi bhi value null ho
# df_cleaned = df.dropna()
# print("Rows after dropping nulls:", df_cleaned.shape)
```
