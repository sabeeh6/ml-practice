```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import seaborn as sns
```

```python
df = pd.read_csv("/content/iris_100_rows.csv")
df
```

```python
df['Species'].unique()
```

```python
df['Species'].replace({'Setosa':'1','Versicolor':'2','Virginica':'3'},inplace=True)
df
```

```python
X_train, X_test, y_train, y_test = train_test_split(df[['SepalLengthCm',	'SepalWidthCm',	'PetalLengthCm',	'PetalWidthCm',]], df['Species'], test_size=0.2, random_state=42)
```

```python
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
prediction = model.predict(X_test)
prediction
```

```python
accuracy= accuracy_score(y_test, prediction)
accuracy
```

```python
model.score(X_train, y_train)
```
